```python
import os
from groq import Groq
import json # Added import for json module

# 1. Setup Client
from google.colab import userdata
api_key =userdata.get("groq")

model_name =  "llama-3.1-8b-instant"

if not api_key:
    raise ValueError("GROQ_API_KEY environment variable is not set.")

client = Groq(api_key=api_key)

def paraphrase_section(text_content):
    """
    Rephrases content by shifting word placement and structure.
    """

    # The System Prompt enforces the "Same Meaning, Different Structure" constraint
    system_instruction = """
    You are a writing assistant focused on 'Rephrasing'.

    YOUR INSTRUCTIONS:
    1. Rewrite the following text using different sentence structures or synonyms.
    2. STRICT CONSTRAINT: The core meaning must remain EXACTLY the same.
    3. STRICT CONSTRAINT: Do NOT add new information or remove details.
    4. Keep the complexity level the same (do not make it overly simple or overly academic).
    5. Output ONLY the paraphrased text, no intro or outro.
    """

    user_message = f"Paraphrase this text:\n\n{text_content}"

    try:
        chat_completion = client.chat.completions.create(
            messages=[
                {"role": "system", "content": system_instruction},
                {"role": "user", "content": user_message}
            ],
            model=model_name,
            temperature=0.5, # Slightly higher temperature for variety in wording
        )
        result_text = chat_completion.choices[0].message.content.strip()

        # Return structured data directly
        return {
            "status": "success",
            "original_text": text_content,
            "paraphrased_text": result_text
        }

    except Exception as e:
        # Return error structure so the API doesn't crash
        return {
            "status": "error",
            "original_text": text_content,
            "error_message": str(e)
        }
print(paraphrase_section("This decision became a landmark precedent for student free speech rights in the United States."))
```
