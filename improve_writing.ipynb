```python
import os
from groq import Groq
import json

# 1. Setup Client using Environment Variables
# Ensure you have set GROQ_API_KEY and GROQ_MODEL in your environment
from google.colab import userdata
api_key =userdata.get("groq")

model_name =  "llama-3.1-8b-instant"

if not api_key:
    raise ValueError("GROQ_API_KEY environment variable is not set.")

client = Groq(api_key=api_key)

def improve_writing_section(text_content):
    """
    Improves clarity and tone without drastic rewrites.
    """

    # The System Prompt enforces the "Subtle Edit" constraint
    system_instruction = """
    You are a professional editor focused on 'Light Polishing'.

    YOUR INSTRUCTIONS:
    1. Improve the grammar, clarity, and flow of the user's text.
    2. STRICT CONSTRAINT: Do NOT change the meaning or the key vocabulary.
    3. STRICT CONSTRAINT: Do NOT drastically change the length. The output word count should be same and maximum 10 words higher than content of the input.
    4. Do NOT sound like an AI; keep the user's original voice, just polished.
    5. Output ONLY the improved text, no intro or outro.
    """

    user_message = f"Here is the text to polish:\n\n{text_content}"

    try:
        chat_completion = client.chat.completions.create(
            messages=[
                {"role": "system", "content": system_instruction},
                {"role": "user", "content": user_message}
            ],
            model=model_name,
            temperature=0.3, # Low temperature to prevent hallucinations/creativity
        )
        improved_text = chat_completion.choices[0].message.content.strip()
        return json.dumps({"improved_text": improved_text})

    except Exception as e:
        return json.dumps({"error": f"Error improving text: {str(e)}"})

# --- Usage Example ---
if __name__ == "__main__":
    sample_text = "i think the project is good but we need to check the data becuase it looks wierd sometimes."
    print( sample_text)
    print( improve_writing_section(sample_text))
```
